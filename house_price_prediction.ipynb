```python
import numpy as np

def split_data(data,test_ratio):
  np.random.seed(42)
  shuffled = np.random.permutation(len(data))
  test_size = int(len(data)*test_ratio)
  test_indices = shuffled[:test_size]
  train_indices = shuffled[test_size:]
  return data.iloc[train_indices], data.iloc[test_indices]
```

```python
train_data, test_data = split_data(data,0.2)
```

```python
print(f"Rows in training data: {len(train_data)}")
print(f"Rows in testing data: {len(test_data)}")
```

```python
train_data_x = train_data.drop('MEDV',axis=1)
train_data_y = train_data['MEDV'].copy()
```

##Creating Pipeline

```python
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler
#from sklearn.impute import simpleimputer
my_pipeline = Pipeline([
    # ('imputer', simpleimputer(startegy='median')), ...add as many paramaters needed in pipeline
    ('std_scaler', StandardScaler())
    ])
```

```python
train_data_prepared = my_pipeline.fit_transform(train_data_x)
```

```python
train_data_prepared.shape # Numpy array
```

```python
from sklearn.linear_model import LinearRegression
from sklearn.tree import DecisionTreeRegressor
from sklearn.ensemble import RandomForestRegressor
dtr = DecisionTreeRegressor()
rfr = RandomForestRegressor()
lin_reg = LinearRegression()
lin_reg.fit(train_data_prepared, train_data_y)
dtr.fit(train_data_prepared, train_data_y)
rfr.fit(train_data_prepared, train_data_y)
```

```python
sample_data = train_data_x.iloc[:5]
sample_data_label = train_data_y.iloc[:5]
```

```python
sample_data_label
```

```python
sample_data_check = my_pipeline.transform(sample_data)
```

```python
lin_reg.predict(sample_data_check)
```

```python
list(sample_data_label)
```

```python
from sklearn.metrics import mean_squared_error
housing_predictions = lin_reg.predict(train_data_prepared)
lin_mse = mean_squared_error(train_data_y, housing_predictions)
lin_rmse = np.sqrt(lin_mse)
```

```python
lin_rmse
```

**Decission tree**

```python
dtr.predict(sample_data_check)
```

```python
from sklearn.metrics import mean_squared_error
housing_predictions_dtr = dtr.predict(train_data_prepared)
lin_mse = mean_squared_error(train_data_y, housing_predictions_dtr)
lin_rmse = np.sqrt(lin_mse)
```

```python
lin_rmse
```

0 RMSE indicates data is overfitted, so we use Cross validation Technique

```python
from sklearn.model_selection import cross_val_score
scores = cross_val_score(dtr, train_data_prepared, train_data_y, scoring='neg_mean_squared_error', cv=10)
rmse_scores_dtr = np.sqrt(-scores)
```

```python
rmse_scores_dtr
```

```python
# doing it for Liner regressor
from sklearn.model_selection import cross_val_score
scores = cross_val_score(lin_reg, train_data_prepared, train_data_y, scoring='neg_mean_squared_error', cv=10)
rmse_scores_lin = np.sqrt(-scores)
```

```python
rmse_scores_lin
```

```python
def print_scores(scores):
  print("Scores: ", scores)
  print("Mean: ", scores.mean())
  print("Standard Deviation: ", scores.std())
```

```python
print("Decision Tree")
print_scores(rmse_scores_dtr)
print("\nLinear Regression")
print_scores(rmse_scores_lin)
```
