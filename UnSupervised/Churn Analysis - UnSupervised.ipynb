```python
from sklearn.decomposition import PCA
pca = PCA(n_components=10)
pca_data = pca.fit_transform(sc_data)
```

```python
VR= pca.explained_variance_ratio_
VR
```

```python
plt.plot(np.cumsum(VR))
```

```python
from sklearn.cluster import KMeans
best_cluster_size = []
for i in range(1,12):
    kmeans = KMeans(n_clusters=i,init='k-means++',random_state=42)
    kmeans.fit(pca_data)
    best_cluster_size.append(kmeans.inertia_)
display(best_cluster_size)
```

```python
plt.plot(range(1,12),best_cluster_size)
```

```python
from sklearn.metrics import silhouette_score
scores =[]
for i in range(2,12):
  k_means= KMeans(n_clusters=i,random_state=42)
  target = k_means.fit_predict(pca_data)
  score = silhouette_score(pca_data,target)
  scores.append(score)
display(scores)
```

```python
plt.figure(figsize=(10,5))

plt.plot(range(2,12), scores, marker='o')


plt.show()
```

```python
model_k = KMeans(n_clusters = 4, random_state=42)
y_label = model_k.fit_predict(pca_data)
```

```python
df1['Target'] = y_label
```

```python
cluster_profile = df1.groupby('Target').mean()
pd.set_option('display.max_columns', None)
cluster_profile
```

```python
risk_assess_features = ["Inactivity_Days",'Customer_Satisfaction',"Transactions_Count",
                'Login_Frequency','Complaints_Count',"Age","Tenure_Months"]
```

```python
from sklearn.preprocessing import MinMaxScaler
mmc = MinMaxScaler()
risk_assess_features_scaled = pd.DataFrame(mmc.fit_transform(df1[risk_assess_features]), columns = risk_assess_features)
```

```python
risk_assess_features_scaled.head()
```

```python
risk_assess_features_scaled.describe()
```

```python
risk_assess_features_scaled["Risk_Score"] = (

    risk_assess_features_scaled["Inactivity_Days"] * 0.35 +

    risk_assess_features_scaled["Complaints_Count"] * 0.25 +

    (1 - risk_assess_features_scaled["Customer_Satisfaction"]) * 0.15 +

    (1 - risk_assess_features_scaled["Transactions_Count"]) * 0.10 +

    (1 - risk_assess_features_scaled["Login_Frequency"]) * 0.10 +

    (1 - risk_assess_features_scaled["Tenure_Months"]) * 0.05
)
```

```python
risk_assess_features_scaled.head()
```

```python
risk_assess_features_scaled["Risk_Score"].describe()
```

```python
risk_assess_features_scaled["Risk_Score"].hist()
```

```python
conditions = [
    risk_assess_features_scaled["Risk_Score"] >= 0.6,
    risk_assess_features_scaled["Risk_Score"] >= 0.4]
choices = [ "High Risk","Moderate Risk"]

df1["Risk_Profile?"] = np.select(conditions, choices, default="Low Risk")
```

```python
df1['Risk_Profile?'].value_counts()
```

```python
pd.crosstab(df1['Cluster'],df1['Risk_Profile?'])
```

We could see cluster 0 has All high risk and majority of moderate risk

```python
df1['Target'] = np.where(df1['Risk_Profile?'] == 'High Risk', 1, 0)
```
